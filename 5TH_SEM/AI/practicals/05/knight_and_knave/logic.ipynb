# **KNIGHT AND KNAVE**

```python
knight = True
knave = False
characters = ['A', 'B']
```

```python
import itertools

for roles in itertools.product([knight, knave], repeat=len(characters)):
    
    is_A_knight, is_B_knight = roles
    statement_A = (not is_A_knight) or (is_B_knight)
    statement_B = is_A_knight
    if (is_A_knight == statement_A) and (is_B_knight == statement_B):
        print("✅ Solution found!")
        print(f"  - A is a {'Knight' if is_A_knight else 'Knave'}")
        print(f"  - B is a {'Knight' if is_B_knight else 'Knave'}")
        break 
else:
    print("❌ No consistent solution found for this puzzle.")
```
