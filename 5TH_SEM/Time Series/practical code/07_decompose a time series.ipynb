# **Decompose a time series**

```python
import pandas as pd
import matplotlib.pyplot as plt
import statsmodels.api as sm
from statsmodels.tsa.seasonal import seasonal_decompose
```

```python
# --- 1. Load the Dataset ---
# We use AirPassengers: Monthly airline passengers 1949-1960
dataset = sm.datasets.get_rdataset("AirPassengers")
df = dataset.data
```

```python
# Set a proper datetime index
df['time'] = pd.date_range(start='1949-01-01', periods=len(df), freq='M')
df.set_index('time', inplace=True)
ts_data = df['value']
```

```python
# --- 2. Perform Additive Decomposition ---
# Ideally, we want: Observed = Trend + Seasonal + Residual
decomposition = seasonal_decompose(ts_data, model='additive')
```

```python
# --- 3. Plot the Components ---
fig = decomposition.plot()
fig.set_size_inches(12, 10)
plt.suptitle('Additive Decomposition of AirPassengers', fontsize=16)
plt.tight_layout()
plt.show()
```
